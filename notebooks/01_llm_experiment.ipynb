# Llama / Gemma 新規実験
Colab GPUカーネルで利用します。最初は設定確認のみです。`RUN = True`にすると指定した1実験を実行します。過去のLoRAセクションはexperiments内の各ノートを参照。

```python
from pathlib import Path
import sys, subprocess
candidates = [Path.cwd(), *Path.cwd().parents, Path('/content/RMT_utils')]
ROOT = next(p for p in candidates if (p / 'notebook_runtime.py').is_file())
sys.path.insert(0, str(ROOT))
from notebook_runtime import output_dir
RUN = False
MODEL = 'llama'  # llama / llama-instruct / gemma
STAGE = 'ppl'  # ppl / esd / lra
RUN_NAME = 'llama_baseline_001'
METRICS = None  # LRA: this runner's esd_metrics.pkl for the same model
MAX_LAYERS = 1
```

```python
command = [sys.executable, str(ROOT / 'tools/run_llm_experiment.py'),
           '--model', MODEL, '--stage', STAGE, '--output', str(output_dir() / RUN_NAME),
           '--max-layers', str(MAX_LAYERS)]
if METRICS is not None:
    command += ['--metrics', str(METRICS)]
print(' '.join(command))
if RUN:
    subprocess.run(command, check=True)
else:
    print('Configuration only. Select a Colab GPU kernel, set a persistent output path, then set RUN = True.')
```
